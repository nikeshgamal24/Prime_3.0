```python
# import the json
import json
```

```python
#load the data from the json file
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)

    return data
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
print(len(data))
```

## Clean Data

```python
def clean_data(data):
    text_to_num = {'one':1, "two":2,"three":3,"four":4,"five":5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        #clean data: handling the mixed type
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating

        #Handling Missing Data
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        #handling duplicates: Deduplication
        if(user["name"] in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)
    return cleaned_data
```

```python
cleaned_data = clean_data(data)
print("NEW DATA")
print(cleaned_data)
print(len(cleaned_data))
print("-"*50)
```

## Meaningful insights
### 1. What is the average rating?
### 2. What % of users gave poor rating(<3) ?

```python
def get_insights(data):
    #calculating the average rating
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])

        # % of users who gave poor rating (<3)
    print(f"Average ratings: {total_rating/len(data)}")
    
    poor_rating = 0
    for user in data:
        if (float(user["rating"]) < 3):
            poor_rating +=1
    print(f"% users who gave poor rating is: {poor_rating/len(data) *100}%")
```

```python
get_insights(cleaned_data)
```
